```python
class NewVisitor:
    name=None
    phone=None
    remarks=None   #None means it is empty now we will assingn in it later


    #constructor
    def _init_(a):
        print("new object created....")
    def my(a):
        print(a.phone)
    def getname(a):
        print(a.name)
    def updatename(a,newname):#one for address and one for name change
        a.name=newname
```

```python
eric=NewVisitor()
```

```python
eric.name="Eric"
```

```python
eric.name
```

```python
eric.phone=9837498274
```

```python
eric.remarks="okay"
```
